```python
# networth calculator

list_asset = []
list_liability = []

# asset
asset = int(input("How many asset you have?: "))
for i in range (asset):
  cost_asset = int(input(f"Enter cost of asset{i+1}: "))
  list_asset.append(cost_asset)
print(list_asset)
asset = sum(list_asset)
print("Sum of all given assets: ",asset)

# liability
print("\n")
liability = int(input("How many laiblities you have?: "))
for i in range(liability):
  cost_liability = int(input(f"Enter cost of liability{i+1}: "))
  list_liability.append(cost_liability)
print(list_liability)
liability = sum(list_liability)
print("Sum of all given liablities: ",liability)

# networth
print("\n")
networth = asset - liability
print("Your networth is: ",networth)
```
