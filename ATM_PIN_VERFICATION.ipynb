```python
tries = 1
while tries<=3:
  pin = int(input("Enter pin here: "))
  if pin == 12345:
   print("Account opened")
   amount = 5000000
   need = input("Do you want to deposit or withdraw: ").lower()
   if need == "deposit":
    deposit = int(input("Enter amount that you want to deposit: "))
    amount = amount + deposit
    print("The amount in your account after depositing is: ",amount)
   elif need == "withdraw":
    withdraw = int(input("Enter amount that you want to withdraw: "))
    amount = amount - withdraw
    print("The amount in your account after withdrawing is: ",amount)
   else:
    print("Calculation went wrong!!! Try again Later")
   break
  else:
   print("Incorrect PIN!!!")
   tries = tries + 1
while tries>3 and pin != 12345:
  print("Acoount Locked!!")
  break
```
