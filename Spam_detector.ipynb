```python
# spam detector
spam_list = ["buy now","click this to get prize","click this","subscribe this","make lot of money","you've won a free prize","we've noticed some suspicious activity on your account. please contact us to resolve"]
text = input("Enter text here: ").lower()

is_spam = any(phrase in text for phrase in spam_list)

if is_spam:
  print(f"{text} might spam. Please be carefull!!!\a\a\a\a")
else:
  print(f"{text} not looks spam.")
```
